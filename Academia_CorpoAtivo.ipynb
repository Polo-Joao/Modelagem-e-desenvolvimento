```python
%%writefile Main.java
import java.util.ArrayList;
import java.util.List;


enum EstadoPlano {
    ATIVO, VENCIDO
}

// Classe Plano encapsulada
class Plano {
    private double valor;
    private EstadoPlano estado;

    public Plano(double valor) {
        this.valor = valor;
        this.estado = EstadoPlano.ATIVO;
    }

    public double getValor() {
        return this.valor;
    }

    public EstadoPlano getEstado() {
        return this.estado;
    }

    public void vencer() {
        this.estado = EstadoPlano.VENCIDO;
    }

    public void renovar() {
        this.estado = EstadoPlano.ATIVO;
    }
}

class CheckIn {
}

// Classe abstrata Aluno
abstract class Aluno {
    private String nome;
    private String documento;
    private Plano plano;
    private List<CheckIn> checkIns;

    public Aluno(String nome, String documento, Plano plano) {
        this.nome = nome;
        this.documento = documento;
        this.plano = plano;
        this.checkIns = new ArrayList<>();
    }

    public String getNome() {
        return nome;
    }

    public Plano getPlano() {
        return plano;
    }

    public List<CheckIn> getCheckIns() {
        return checkIns;
    }

    public abstract double valorMensalidade();

    public void realizarCheckIn() {
        if (plano.getEstado() == EstadoPlano.VENCIDO) {
            throw new IllegalStateException("Não é permitido realizar check-in com plano vencido.");
        }
        checkIns.add(new CheckIn());
    }
}

// AlunoComum: paga o valor integral do plano
class AlunoComum extends Aluno {
    public AlunoComum(String nome, String documento, Plano plano) {
        super(nome, documento, plano);
    }

    @Override
    public double valorMensalidade() {
        return getPlano().getValor();
    }
}

// AlunoPlanoFamilia: divide o valor do plano entre o titular e dependentes
class AlunoPlanoFamilia extends Aluno {
    private int quantidadeDependentes;

    public AlunoPlanoFamilia(String nome, String documento, Plano plano, int quantidadeDependentes) {
        super(nome, documento, plano);
        this.quantidadeDependentes = quantidadeDependentes;
    }

    public int getQuantidadeDependentes() {
        return quantidadeDependentes;
    }

    @Override
    public double valorMensalidade() {
        // Divide o valor entre o titular + dependentes
        return getPlano().getValor() / (1 + quantidadeDependentes);
    }
}

// Turma que agrega Alunos
class Turma {
    private String nome;
    private List<Aluno> alunos;

    public Turma(String nome) {
        this.nome = nome;
        this.alunos = new ArrayList<>();
    }

    public String getNome() {
        return nome;
    }

    public List<Aluno> getAlunos() {
        return alunos;
    }

    public void adicionarAluno(Aluno aluno) {
        alunos.add(aluno);
    }

    public void removerAluno(Aluno aluno) {
        alunos.remove(aluno);
    }
}

// Classe Principal de Testes
public class Main {
    public static void main(String[] args) {
        System.out.println("==========================================");
        System.out.println("   ACADEMIA CORPORATIVO - TESTES UML");
        System.out.println("==========================================\n");

        // =========================================================
        // 1. ABSTRAÇÃO
        // =========================================================
        System.out.println("1. ABSTRAÇÃO");
        System.out.println("------------------------------------------");
        System.out.println("✓ Aluno é uma classe abstrata.");
        System.out.println("✓ Não pode ser instanciada diretamente.");
        System.out.println();

        // =========================================================
        // 2. HERANÇA
        // =========================================================
        System.out.println("2. HERANÇA");
        System.out.println("------------------------------------------");

        Plano planoJoao = new Plano(150.00);
        Plano planoMaria = new Plano(300.00);

        AlunoComum joao = new AlunoComum("João", "12345678900", planoJoao);
        AlunoPlanoFamilia maria = new AlunoPlanoFamilia("Maria", "98765432100", planoMaria, 3);

        System.out.println("AlunoComum é um Aluno: " + (joao instanceof Aluno));
        System.out.println("AlunoPlanoFamilia é um Aluno: " + (maria instanceof Aluno));
        System.out.println("✓ As duas classes herdam de Aluno.");
        System.out.println();

        // =========================================================
        // 3. ENCAPSULAMENTO
        // =========================================================
        System.out.println("3. ENCAPSULAMENTO");
        System.out.println("------------------------------------------");
        System.out.println("Valor inicial do plano de João: R$ " + String.format("%.2f", planoJoao.getValor()));
        System.out.println("Estado inicial: " + planoJoao.getEstado());

        planoJoao.vencer();
        System.out.println("Estado após vencer(): " + planoJoao.getEstado());
        System.out.println("✓ Estado e valor do plano são encapsulados.");
        System.out.println("✓ O estado é alterado através de operações controladas.");
        System.out.println();

        // =========================================================
        // 4. POLIMORFISMO
        // =========================================================
        System.out.println("4. POLIMORFISMO");
        System.out.println("------------------------------------------");
        Aluno aluno1 = joao;
        Aluno aluno2 = maria;

        System.out.println("Mensalidade do AlunoComum: R$ " + String.format("%.2f", aluno1.valorMensalidade()));
        System.out.println("Mensalidade do AlunoPlanoFamilia: R$ " + String.format("%.2f", aluno2.valorMensalidade()));
        System.out.println("✓ O mesmo método valorMensalidade() possui comportamentos diferentes.");
        System.out.println();

        // =========================================================
        // 5. ASSOCIAÇÃO
        // =========================================================
        System.out.println("5. ASSOCIAÇÃO: ALUNO -> PLANO");
        System.out.println("------------------------------------------");
        System.out.println("João possui o plano: " + joao.getPlano().getEstado());
        System.out.println("Maria possui o plano: " + maria.getPlano().getEstado());
        System.out.println("✓ Cada aluno possui 1 plano ativo.");
        System.out.println();

        // =========================================================
        // 6. COMPOSIÇÃO
        // =========================================================
        System.out.println("6. COMPOSIÇÃO: ALUNO ◆── CHECKIN");
        System.out.println("------------------------------------------");
        System.out.println("Tentando fazer check-in com plano vencido...");
        try {
            joao.realizarCheckIn();
            System.out.println("Check-in realizado.");
        } catch (IllegalStateException e) {
            System.out.println("ERRO: " + e.getMessage());
        }
        System.out.println();

        System.out.println("Renovando o plano de João...");
        joao.getPlano().renovar();
        System.out.println("Novo estado: " + joao.getPlano().getEstado());

        joao.realizarCheckIn();
        joao.realizarCheckIn();
        System.out.println("Quantidade de check-ins de João: " + joao.getCheckIns().size());
        System.out.println("✓ CheckIns pertencem ao ciclo de vida do Aluno.");
        System.out.println("✓ A criação do CheckIn é controlada pelo Aluno.");
        System.out.println();

        // =========================================================
        // 7. REGRA DE NEGÓCIO / INVARIANTE
        // =========================================================
        System.out.println("7. INVARIANTE DO SISTEMA");
        System.out.println("------------------------------------------");
        System.out.println("Vencendo o plano de Maria...");
        maria.getPlano().vencer();
        System.out.println("Estado do plano: " + maria.getPlano().getEstado());
        System.out.println("Tentando fazer check-in...");
        try {
            maria.realizarCheckIn();
            System.out.println("Check-in realizado.");
        } catch (IllegalStateException e) {
            System.out.println("ERRO: " + e.getMessage());
        }
        System.out.println("✓ Um aluno não pode realizar check-in com plano vencido.");
        System.out.println();

        // =========================================================
        // 8. AGREGAÇÃO
        // =========================================================
        System.out.println("8. AGREGAÇÃO: TURMA ◇── ALUNO");
        System.out.println("------------------------------------------");
        Turma musculacao = new Turma("Musculação");
        musculacao.adicionarAluno(joao);
        musculacao.adicionarAluno(maria);

        System.out.println("Turma: " + musculacao.getNome());
        System.out.println("Quantidade de alunos: " + musculacao.getAlunos().size());
        System.out.println("\nAlunos da turma:");
        for (Aluno aluno : musculacao.getAlunos()) {
            System.out.println("- " + aluno.getNome());
        }
        System.out.println();

        musculacao.removerAluno(joao);
        System.out.println("João foi removido da turma.");
        System.out.println("João ainda existe no sistema: " + joao.getNome());
        System.out.println("Quantidade de check-ins de João: " + joao.getCheckIns().size());
        System.out.println("✓ O Aluno continua existindo independentemente da Turma.");
        System.out.println("✓ Por isso a relação é uma agregação.");
        System.out.println();

        // =========================================================
        // 9. MULTIPLICIDADE
        // =========================================================
        System.out.println("9. MULTIPLICIDADES");
        System.out.println("------------------------------------------");
        System.out.println("Aluno -> Plano: 1");
        System.out.println("Aluno -> CheckIn: *");
        System.out.println("Turma -> Aluno: *");
        System.out.println("✓ As relações possuem multiplicidades conforme o UML.");
        System.out.println();

        // =========================================================
        // 10. TESTE DO PLANO FAMILIA
        // =========================================================
        System.out.println("10. CÁLCULO DO PLANO FAMÍLIA");
        System.out.println("------------------------------------------");
        System.out.println("Valor total do plano: R$ " + String.format("%.2f", maria.getPlano().getValor()));
        System.out.println("Quantidade de dependentes: " + maria.getQuantidadeDependentes());
        System.out.println("Mensalidade por pessoa: R$ " + String.format("%.2f", maria.valorMensalidade()));
        System.out.println("✓ O plano família divide o valor entre o titular e os dependentes.");
        System.out.println();

        // =========================================================
        // RESULTADO FINAL
        // =========================================================
        System.out.println("==========================================");
        System.out.println("          TESTES FINALIZADOS");
        System.out.println("==========================================");
    }
}
```
